```python
#1 MB = 1024 kb

#Code define --> agr item me -1 pe M hai to use replace kro empty string se then uske convert kro number me(float) me
def size_process(item):
    if str(item)[-1] == "M":
        res = float(str(item).replace("M", ""))
        res = res * 1024
        return res
    elif str(item)[-1] == "k":
        res = float(str(item).replace("k", ""))
        res = res * 1024
        return res
    else:
        return str(np.NaN)
```

```python
# Internal assignment>> prepare Android Ver remove ( -, w, 3rd decimal no ex: 7.3.3 so remove last decimal like this 7.3 )
```

```python
def clean_android_version(version):
    if not isinstance(version, str) or version.strip() == "" or pd.isna(version):
        return ""  # NaN aur empty values ko "" banayenge
    
    # Remove hyphen (take first part before "-")
    version = version.split(" - ")[0]
    
    # Remove 'W' (replace with empty string)
    version = version.replace("W", "")

    # Split by '.' and take only first two parts
    parts = version.strip().split(".")
    
    # Keep only 'x.y' (first two parts) if three exist
    return ".".join(parts[:2]) if len(parts) > 1 else version
```

```python
df_copy["Android Ver"] = df_copy["Android Ver"].apply(clean_android_version)
```
